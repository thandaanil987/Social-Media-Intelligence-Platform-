```python
import pandas as pd
import numpy as np

# Load datasets
accounts = pd.read_csv("Dataset/accounts.csv")
campaigns = pd.read_csv("Dataset/campaigns.csv")
content = pd.read_csv("Dataset/content_performance.csv")
audience = pd.read_csv("Dataset/audience_growth.csv")
```

```python
# Store datasets in a dictionary
datasets = {
    "Accounts": accounts,
    "Campaigns": campaigns,
    "Content Performance": content,
    "Audience Growth": audience
}

# Check basic information
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    
    print("\nShape:", df.shape)
    
    print("\nColumn Names:")
    print(df.columns.tolist())
    
    print("\nData Types:")
    print(df.dtypes)
    
    print("\nMissing Values:")
    print(df.isnull().sum())
    
    print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
for name, df in datasets.items():
    print(f"\n{'=' * 60}")
    print(name)
    display(df.head())
```

```python
# Convert date columns to datetime format

campaigns["start_date"] = pd.to_datetime(campaigns["start_date"])
campaigns["end_date"] = pd.to_datetime(campaigns["end_date"])

content["post_date"] = pd.to_datetime(content["post_date"])

audience["date"] = pd.to_datetime(audience["date"])
```

```python
# Check missing values in all datasets

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())
```

```python
# Check duplicate rows

for name, df in datasets.items():
    print(f"\n{name}: {df.duplicated().sum()} duplicate rows")
```

```python
# Check numeric data quality

data_quality_checks = {
    "Accounts": (accounts, ["followers_start"]),
    
    "Campaigns": (campaigns, ["budget"]),
    
    "Content Performance": (
        content,
        [
            "reach", "impressions", "likes",
            "comments", "shares", "saves",
            "clicks", "followers_gained"
        ]
    ),
    
    "Audience Growth": (
        audience,
        [
            "followers_gained",
            "followers_lost",
            "followers_end"
        ]
    )
}

for name, (df, columns) in data_quality_checks.items():
    print(f"\n{'='*50}")
    print(name)
    print(df[columns].describe())
```

```python
# Calculate campaign duration in days
campaigns["campaign_duration_days"] = (
    campaigns["end_date"] - campaigns["start_date"]
).dt.days
```

```python
# Calculate total engagement
content["total_engagement"] = (
    content["likes"]
    + content["comments"]
    + content["shares"]
    + content["saves"]
)
```

```python
# Calculate engagement rate (%)
content["engagement_rate"] = (
    content["total_engagement"] / content["reach"] * 100
).round(2)
```

```python
# Calculate net follower growth
audience["net_follower_growth"] = (
    audience["followers_gained"] - audience["followers_lost"]
)
```

```python
import sqlite3

# Create SQLite database connection
conn = sqlite3.connect("social_media.db")

print("SQLite database connected successfully!")
```

```python
# Load prepared datasets into SQLite

accounts.to_sql(
    "accounts",
    conn,
    if_exists="replace",
    index=False
)

campaigns.to_sql(
    "campaigns",
    conn,
    if_exists="replace",
    index=False
)

content.to_sql(
    "content_performance",
    conn,
    if_exists="replace",
    index=False
)

audience.to_sql(
    "audience_growth",
    conn,
    if_exists="replace",
    index=False
)

print("All tables loaded successfully!")
```

```python
query = """
SELECT name 
FROM sqlite_master 
WHERE type='table';
"""

tables = pd.read_sql_query(query, conn)
display(tables)
```

```python
for table in tables["name"]:
    count_query = f"SELECT COUNT(*) AS total_records FROM {table}"
    result = pd.read_sql_query(count_query, conn)
    print(f"{table}: {result['total_records'][0]} records")
```

```python
# Overall Performance
query = """
SELECT
    COUNT(*) AS total_posts,
    SUM(reach) AS total_reach,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate
FROM content_performance;
"""

pd.read_sql_query(query, conn)
```

```python
# Platform Performance
query = """
SELECT
    platform,
    COUNT(*) AS total_posts,
    SUM(reach) AS total_reach,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate
FROM content_performance
GROUP BY platform
ORDER BY total_reach DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Content Type Performance
query = """
SELECT
    content_type,
    COUNT(*) AS total_posts,
    SUM(reach) AS total_reach,
    ROUND(AVG(total_engagement), 0) AS avg_engagement,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate
FROM content_performance
GROUP BY content_type
ORDER BY avg_engagement_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Topic Performance

query = """
SELECT
    topic,
    COUNT(*) AS total_posts,
    SUM(reach) AS total_reach,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate
FROM content_performance
GROUP BY topic
ORDER BY avg_engagement_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Campaign Performance Analysis
query = """
SELECT
    c.campaign_name,
    c.platform,
    c.budget,
    c.objective,
    COUNT(cp.post_id) AS total_posts,
    COALESCE(SUM(cp.reach), 0) AS total_reach,
    ROUND(AVG(cp.engagement_rate), 2) AS avg_engagement_rate
FROM campaigns c
LEFT JOIN content_performance cp
    ON c.campaign_id = cp.campaign_id
GROUP BY
    c.campaign_id,
    c.campaign_name,
    c.platform,
    c.budget,
    c.objective
ORDER BY total_reach DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# top Campaigns by Engagement
query = """
SELECT
    c.campaign_name,
    c.objective,
    COUNT(cp.post_id) AS total_posts,
    ROUND(AVG(cp.engagement_rate), 2) AS avg_engagement_rate,
    SUM(cp.total_engagement) AS total_engagement
FROM campaigns c
JOIN content_performance cp
    ON c.campaign_id = cp.campaign_id
GROUP BY
    c.campaign_id,
    c.campaign_name,
    c.objective
ORDER BY avg_engagement_rate DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# Account-wise Audience Growth
query = """
SELECT
    a.account_name,
    a.platform,
    SUM(ag.followers_gained) AS total_followers_gained,
    SUM(ag.followers_lost) AS total_followers_lost,
    SUM(ag.net_follower_growth) AS net_follower_growth
FROM accounts a
JOIN audience_growth ag
    ON a.account_id = ag.account_id
GROUP BY
    a.account_id,
    a.account_name,
    a.platform
ORDER BY net_follower_growth DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Monthly Audience Growth Trend
query = """
SELECT
    date,
    SUM(followers_gained) AS followers_gained,
    SUM(followers_lost) AS followers_lost,
    SUM(net_follower_growth) AS net_growth
FROM audience_growth
GROUP BY date
ORDER BY date;
"""

pd.read_sql_query(query, conn)
```

```python
# Top 10 Best Performing Posts
query = """
SELECT
    post_id,
    platform,
    content_type,
    topic,
    reach,
    total_engagement,
    engagement_rate
FROM content_performance
ORDER BY engagement_rate DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# Best Content Type by Platform
query = """
SELECT
    platform,
    content_type,
    COUNT(*) AS total_posts,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate,
    SUM(reach) AS total_reach
FROM content_performance
GROUP BY platform, content_type
ORDER BY platform, avg_engagement_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Best Topic by Platform
query = """
SELECT
    platform,
    topic,
    COUNT(*) AS total_posts,
    SUM(reach) AS total_reach,
    ROUND(AVG(engagement_rate), 2) AS avg_engagement_rate
FROM content_performance
GROUP BY platform, topic
ORDER BY platform, avg_engagement_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# Account Performance Summary

query = """
SELECT
    a.account_name,
    a.platform,
    a.category,
    COUNT(cp.post_id) AS total_posts,
    SUM(cp.reach) AS total_reach,
    SUM(cp.total_engagement) AS total_engagement,
    ROUND(AVG(cp.engagement_rate), 2) AS avg_engagement_rate
FROM accounts a
JOIN content_performance cp
    ON a.account_id = cp.account_id
GROUP BY
    a.account_id,
    a.account_name,
    a.platform,
    a.category
ORDER BY total_reach DESC;
"""

pd.read_sql_query(query, conn)
```
